Tokenization

```python
import re

def tokenize(text):
    text = text.lower()                      # lowercase
    text = re.sub(r"[\"']", "", text)        # remove quotes
    text = re.sub(r"[^a-z0-9\s]", " ", text) # remove punctuation
    text = re.sub(r"\s+", " ", text).strip() # replace multiple spaces with one

    tokens = text.split()                    # tokenize into words
    return tokens
```

Vocab Forming

```python
vocab = {'<UNK>':0}
```

```python
def build_voceb (row):

  #print(row['Question'], row['Answer'])

  tokenized_question = tokenize(row['Question'])
  tokenized_answer = tokenize(row['Answer'])

  merged_tokens = tokenized_question + tokenized_answer

  # print (merged_tokens)

  for token in merged_tokens:
    if token not in vocab:
      vocab[token]=len(vocab)
```

```python
df.apply(build_voceb, axis =1)
```

# Text to Indices

```python
def text_to_indices(text,vocab):
  indexed_text = []

  for token in tokenize(text):
    if token in vocab:
      # print(vocab[token])
      indexed_text.append(vocab[token])

    else:
      indexed_text.append(vocab['<UNK>'])

  return indexed_text
```

```python
import torch
from torch.utils.data import Dataset,DataLoader
```

```python
class QADataset(Dataset):

  def __init__(self , df,vocab):
    self.df =df
    self.vocab =vocab
  def __len__(self):
    return self.df.shape[0]
  def __getitem__(self,index):
    numerical_question = text_to_indices (self.df.iloc[index]['Question'],self.vocab)
    numerical_answer = text_to_indices (self.df.iloc[index]['Answer'],self.vocab)

    return torch.tensor(numerical_question), torch.tensor(numerical_answer)
```

```python
dataset = QADataset(df, vocab)
```

```python
dataloader = DataLoader(dataset , batch_size=1 , shuffle=True)
```

# RNN architecture implementation

```python
import torch.nn as nn

class simpleRNN(nn.Module):
  def __init__(self, vocab_size , embedding_dim =50 , hidden_size=64):
    super().__init__()

    self.embedding = nn.Embedding(vocab_size , embedding_dim)
    self.rnn = nn.RNN(embedding_dim , hidden_size, batch_first=True)
    self.fc = nn.Linear(hidden_size, vocab_size)

  def forward (self, Question):
    embedded = self.embedding(Question)
    _,final = self.rnn(embedded)

    return self.fc(final.squeeze(0))
```

```python
model = simpleRNN(vocab_size=len(vocab))
```

# Training loop

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.0001)
epochs =50
```

```python
for epoch in range(epochs):
  total_loss=0
  for Question,Answer in dataloader:
    optimizer.zero_grad()
    output = model(Question)

    target = Answer[0][0].unsqueeze(0)

    loss = criterion(output, target)
    loss.backward()
    optimizer.step()
    total_loss+=loss.item()

  if (epoch + 1) % 10 == 0:
    print(f"Epoch{epoch+1:3d}/{epochs}|Loss:{total_loss:.4f}")
```

# Prediction

```python
vocab.items()
```

```python
idx_to_word = {idx:word for word, idx in vocab.items()}
idx_to_word
```

```python
def predict(question_text , model, vocab , idx_to_word):
  model.eval()

  with torch.no_grad():
    indices = text_to_indices( question_text , vocab)

    if not indices:
      return"<UNK>"

    x= torch.tensor(indices).unsqueeze(0)

    pred = torch.argmax(model(x),dim=1).item()

  return idx_to_word.get(pred,"<UNK>")
```
